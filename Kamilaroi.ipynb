```python
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

# Hyperparameters
VOCAB_SIZE = 2710    # Total unique characters in the language
EMBED_DIM = 64     # How we represent characters as vectors
SEQ_LEN = 10       # Context window size
BATCH_SIZE = 4     # Number of examples processed together
EPOCHS = 25        # Training cycles
```

```python
class LanguageDataset(Dataset):
    def __init__(self, text, seq_len=SEQ_LEN):
        self.chars = sorted(list(set(text)))
        self.vocab = {c:i for i,c in enumerate(self.chars)}
        self.data = [self.vocab[c] for c in text]
        self.seq_len = seq_len
        
    def __len__(self):
        return len(self.data) - self.seq_len
        
    def __getitem__(self, idx):
        x = torch.tensor(self.data[idx:idx+self.seq_len])
        y = torch.tensor(self.data[idx+1:idx+self.seq_len+1])
        return x, y

# Create training data
text = " ".join(words).lower()
dataset = LanguageDataset(text)
dataloader = DataLoader(dataset, batch_size=BATCH_SIZE, shuffle=True)
```

```python
class SimpleLLM(nn.Module):
    def __init__(self):
        super().__init__()
        self.embed = nn.Embedding(VOCAB_SIZE, EMBED_DIM)
        self.transformer = nn.TransformerEncoder(
            nn.TransformerEncoderLayer(d_model=EMBED_DIM, nhead=4),
            num_layers=2
        )
        self.fc = nn.Linear(EMBED_DIM, VOCAB_SIZE)
        
    def forward(self, x):
        x = self.embed(x)
        x = self.transformer(x)
        return self.fc(x)

model = SimpleLLM()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
criterion = nn.CrossEntropyLoss()
```

```python
for epoch in range(EPOCHS):
    for inputs, targets in dataloader:
        outputs = model(inputs)
        loss = criterion(outputs.view(-1, VOCAB_SIZE), targets.view(-1))
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
    print(f"Epoch {epoch+1} Loss: {loss.item():.4f}")
```

```python
def generate(prompt, length=5):
    model.eval()
    with torch.no_grad():
        input_seq = torch.tensor([dataset.vocab[c] for c in prompt.lower()])
        for _ in range(length):
            output = model(input_seq[-SEQ_LEN:].unsqueeze(0))
            next_token = output.argmax(-1)[-1].item()
            input_seq = torch.cat((input_seq, torch.tensor([next_token])))
    return ''.join([dataset.chars[i] for i in input_seq.tolist()])


print(generate("bur"))  # Should output something like "buruma"
```
